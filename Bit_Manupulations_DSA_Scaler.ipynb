# Q1: Given an array where every number appears twice except one → find the unique number.

```python
A = [ 4,5,5,1,6,4,6]
```

```python
# Brute force method

def brute_unique(arr):
  n = len(arr)

  for i in range(n):
    count = 0

    for j in range(n):
      if arr[i] == arr[j]:
        count += 1
    if count == 1:
      return arr[i]
```

```python
brute_unique([ 4,5,5,1,6,4,6])
```

```python
# Time complexity: O(n^2)
# Space complexity: O(1)
```

#Logic Using XOR

x ^ x = 0

0 ^ x = x

So all pairs cancel out, only the single element remains.

```python
def singleNumber(nums):
  result = 0
  for i in nums:
    result ^=i
  return result
```

```python
nums = [2, 1, 4, 2, 1]
singleNumber(nums)
```

```python
# Time complexity: O(n)
# Space complexity: O(1)
```

```python
#check + flip
def unset(A,i):
  if A & (1<<i)>0:
    A= A ^ (1<<i)
  return A
```

# Q3: Count no. of set bits in given interger N.

```python
# for 45, ans=4
# for 5, ans =2
```

```python
# integer will have 32 bits
```

```python
def countbits(n):
  count = 0
  for i in range(32):
    if (n & (1<<i))>0:
      count += 1
  return count
```

```python
countbits(111)
```

```python
countbits(331)
```

```python
# Time comlexity O(32) --> O(1)
# Space complexity O(1)
```

# Q4: Given an integer array where every number occur twice except for two numbers. Find those two numbers.

```python
def twonum(nums):
  res = 0
  for i in nums:
    res ^=i
  return res
```

```python
twonum([4,5,4,1,6,6,5,2])
```

```python
5^6
```

```python
# XOR of all elements --> 1^2 gives: 3
# x^x=0
# x^y  !=0
```

```python
def find_two_unique(A):
  xor_all = 0
  for n in A:
    xor_all ^=n

  k=0
  for i in range(32):
    if xor_all & (1<<i) >0:
      k=i
      break

  x=0
  y=0

  for n in A:
    if n & (1<<k):
      x ^=n
    else:
      y ^=n

  return x,y


```

```python
find_two_unique([2, 4, 7, 9, 2, 4])
```

```python
# Time complexity = O(N + 32 + N) = O(N)
# Space complexity = O(1)
```

STEP-1
🔥 XOR all numbers
xor_all = 0
for n in A:
    xor_all ^= n

Why?

XOR of two identical numbers is 0 (a ^ a = 0)

XOR with 0 gives the same number (x ^ 0 = x)

XOR is commutative and associative

So after XORing the whole array:

xor_all = unique1 ^ unique2


Because all duplicates cancel out.

Example array:

[2, 4, 7, 9, 2, 4]

xor_all = 7 ^ 9

STEP-2
🔥 Find a bit that is different between the two unique numbers

for i in range(32):
    if xor_all & (1 << i):
        K = i
        break

What does (1 << i) do?

It creates a mask with only the i-th bit = 1.

Bit example:

1 << 0 = 0001

1 << 1 = 0010

1 << 2 = 0100

1 << 3 = 1000
...

Why check xor_all & (1 << i)?

Because:

If this bit is set (=1) in xor_all,

→ then the two unique numbers differ at this bit.

Example:

7  = 0111

9  = 1001

xor=1110


Here, bits 1, 2, 3 are set.

Choosing any one (say bit 1 or bit 2 or bit 3) is ok.

Let’s say we found bit K = 1.

This bit helps us separate the two unique numbers.

STEP-3
🔥 Divide the numbers into 2 groups based on bit K
if n & (1 << K):
    x ^= n
else:
    y ^= n

Group 1 (bit K = 1)

Numbers having that bit = 1 go here.

This group contains one unique number.

Duplicates cancel because:

a ^ a = 0

Group 2 (bit K = 0)

Numbers having that bit = 0 go here.
This group contains the other unique number.

Duplicates cancel here also.

So:

XORing group 1 gives unique1

XORing group 2 gives unique2

STEP-4
🎉 Return both unique numbers

return x, y
