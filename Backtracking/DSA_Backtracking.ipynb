# Q1: Given n pairs of parentheses, generate all combinations of well-formed parentheses.

```python
def printAll(res, co, cc, N):

  if co == N and cc == N:
    print(res)
    return

  if co > N:
    return

  if cc > co:
    return

  printAll(res+'(', co+1, cc, N)
  printAll(res+')', co, cc+1, N)
```

```python
printAll('' ,0,0,3)
```

```python
def generateParenthesis(n):
    result = []

    def backtrack(current, open_count, close_count):

        # Base case
        if len(current) == 2 * n:
            result.append(current)
            return

        # Add "(" if we still can
        if open_count < n:
            backtrack(current + "(", open_count + 1, close_count)

        # Add ")" only if valid
        if close_count < open_count:
            backtrack(current + ")", open_count, close_count + 1)

    backtrack("", 0, 0)
    return result
```

```python
generateParenthesis(2)
```

Time = Number of valid combinations × length of each string

Space = Recursion stack O(n) + output storage O(Catalan(n))

```python
def printallsubsequences(arr,index,ans):

  if index == len(arr):
    print(ans)
    return

  # coming
  printallsubsequences(arr,index+1,ans+str(arr[index])+' ')

  # not coming
  printallsubsequences(arr,index+1,ans)
```

```python
arr=[10,20,30]
printallsubsequences(arr,0,'')
```

```python
# 2 nd method

def printSubsequences(arr):

    def backtrack(index, current):

        if index == len(arr):
            print(current)
            return

        # Include current element
        backtrack(index + 1, current + [arr[index]])

        # Exclude current element
        backtrack(index + 1, current)

    backtrack(0, [])
```

```python
printSubsequences([1,2,3])
```

At every index, we have two choices — include or exclude the element.
This forms a recursion tree of size 2^n."

🧠 Even More Professional Version (Using Backtracking Properly)

Avoid creating new list every time:

```python
def printSubsequences(arr):
    result = []

    def backtrack(index, current):

        if index == len(arr):
            result.append(current[:])  # copy
            return

        # Include
        current.append(arr[index])
        backtrack(index + 1, current)

        # Undo (Backtrack)
        current.pop()

        # Exclude
        backtrack(index + 1, current)

    backtrack(0, [])
    return result
```

```python
printSubsequences([1,2,3])
```

We:

Modify list

Revert change using pop()

This is called real backtracking

Time:  O(2**n)

Space: O(n)

# 3Q. Given a string, print all possible permutations of the string using recursion.

```python
def AllPerms(ques, ans):

    # Base case
    if len(ques) == 0:
        print(ans)
        return

    # Loop through each character
    for i in range(len(ques)):

        # Current character
        ch = ques[i]

        # Remaining string after removing current character
        left = ques[0:i]
        right = ques[i+1:]

        remaining = left + right

        # Recursive call
        AllPerms(remaining, ans + ch)


# Function call
AllPerms("abc", "")
```
