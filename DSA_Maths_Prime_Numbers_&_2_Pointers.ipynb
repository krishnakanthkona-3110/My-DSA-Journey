```python
def countfac(p):
  c=0
  for i in range(1,p+1):
    if p%i ==0:
      c +=1
  return c
```

```python
countfac(25)
```

Time complexity - O(n)

```python
def fac(n):
  f=0
  i=1
  while i*i<=n:
    if n%i==0:
      if i==n/i:
        f+=1
      else:
        f +=2
    i +=1
  return f
```

```python
fac(24)
```

Time complexity: O(√n)

Space complexity: O(1)

# Prime Numbers - A prime number is a number:

Greater than 1

Has exactly 2 factors

```python
def printprimes(n):
  for i in range(1,n+1):
    if fac(i)==2:
      print(i)
```

```python
printprimes(10)
```

Time complexity - O(n√n)

# Sieve of Eratosthenes

```python
def sieve(n):
    # Step 1: Assume all numbers are prime
    prime = [True] * (n + 1)

    # Step 2: 0 and 1 are not prime
    prime[0] = prime[1] = False

    # Step 3: Mark multiples of each prime
    i = 2
    while i * i <= n:
        if prime[i]:                 # if i is prime
            j = i * i               # start from i^2
            while j <= n:
                prime[j] = False    # mark composite
                j += i
        i += 1

    # Step 4: Print all primes
    for i in range(2, n + 1):
        if prime[i]:
            print(i)


# Example run
sieve(20)
```

#Qs: Given a sorted array A and integer K,find any pair (i, j)
such that A[i] + A[j] = K and i ≠ j.

Brute Force Approach

Idea:

Check every possible pair

If sum equals K, return that pair

```python
def brutesum(arr,k):
  n = len(arr)
  for i in range(n):
    for j in range(i+1,n):
      if arr[i]+arr[j]==k:
          #return (arr[i],arr[j])
          return (i,j)
  return None
```

```python
brutesum([1,2,3,4],9)
```

Time complexity: O(n^2)

Space complexity: O(1)

```python
def optsum(arr,k):
  n=len(arr)
  hs = set()
  for i in range(n):
    if (k-arr[i]) in hs:
      return True
    else:
      hs.add(arr[i])
  return False
```

```python
optsum([1,2,3,4,5],5)
```

Time complexity: O(n)

Space complexity: O(n)

```python
# Brute force method

def pairs(arr,k):
  n=len(arr)
  res = []
  count = 0
  for i in range(n):
    for j in range(i+1,n):
      if arr[i]+arr[j]==k:
        count +=1
        res.append((arr[i],arr[j]))
  return res, 'pairs-',count
```

```python
pairs([1,2,3,4,5,6],6)
```

```python
pairs([1,2,3,4,5,6,5],6)
```

Time complexity: O(n^2)

Space Complexity:
- The space used for the result list res depends on the number of pairs that sum to k.
- In the worst case, if all pairs sum to k, the number of pairs is O(n^2).
- Apart from the output list, the function uses a constant amount of additional space.
- Therefore, the space complexity is O(n^2) in the worst case.

```python
# case 1: When elements are distinct

def countp(arr,k):
  n = len(arr)
  hs = set()
  count = 0
  for i in range(n):
    if k-arr[i] in hs:
      count +=1
    hs.add(arr[i])
  return count
```

```python
countp([1,2,3,4,5],5)
```

```python
# case 2: When elements are repeating (duplicates)

def pairsumo(arr,k):
  n = len(arr)
  dict = {}
  count=0

  for i in range(n):
    pair = k-arr[i]
    count += dict.get(pair,0)
    dict[arr[i]]=dict.get(arr[i],0) +1
  return count
```

```python
pairsumo([1,2,3,4,5],5)
```

Time complexity: O(n)

Space complexity: O(n)

```python
def brute2_sum(arr,k):
  n = len(arr)

# Outer loop picks the starting index of subarray
  for i in range(n):
    curr_sum = 0
    # Inner loop expands the subarray from index i to end
    for j in range(i,n):
      curr_sum += arr[j]
      if curr_sum == k:
        return True
  return False
```

```python
brute2_sum([1,2,3,4,5,6,7],10)
```

```python
def subarray_sum(arr, k):
    prefix_sum = 0          # Running cumulative sum
    seen = set()            # Stores previous prefix sums

    for i in range(len(arr)):

        prefix_sum += arr[i]   # Add current element

        # Case 1: If prefix sum itself equals k
        if prefix_sum == k:
            return True

        # Case 2: If (prefix_sum - k) seen before
        # That means subarray exists with sum k
        if (prefix_sum - k) in seen:
            return True

        # Store current prefix sum
        seen.add(prefix_sum)

    return False
```

```python
subarray_sum([1,2,3,4,5],5)
```

```python
# Time and space complexity is O(n)
```

⚠ Important Note

Your current function:

return True


It will stop at first occurrence only.

If you want to detect all occurrences or count them,
you need to use a dictionary instead of set.

```python
def count_subarrays(arr, k):
    prefix_sum = 0
    count = 0
    seen = {0: 1}   # important!

    for num in arr:
        prefix_sum += num

        if prefix_sum - k in seen:
            count += seen[prefix_sum - k]

        seen[prefix_sum] = seen.get(prefix_sum, 0) + 1

    return count
```

```python
count_subarrays([1,2,3,2,1],5)
```

🧠 Interview Insight

There are 3 variations:

Check if exists → use set

Count total → use hashmap with frequency

Longest subarray → store first occurrence only

```python
def longest_subarray(arr, k):
    prefix_sum = 0
    max_len = 0
    seen = {}

    for i in range(len(arr)):
        prefix_sum += arr[i]

        # Case 1: whole array from 0 to i
        if prefix_sum == k:
            max_len = i + 1

        # Case 2: prefix_sum - k seen before
        if prefix_sum - k in seen:
            length = i - seen[prefix_sum - k]
            max_len = max(max_len, length)

        # Store only first occurrence
        if prefix_sum not in seen:
            seen[prefix_sum] = i

    return max_len
```

```python
arr = [1, -1, 5, -2, 3]
k = 3
longest_subarray(arr,k)
```
