# 1). Bubble Sort:
Compare adjacent elements and swap if they are in the wrong order. After each pass, the largest element "bubbles" to the end.

image.png

```python
def bubble_sort(arr):
  n = len(arr)

  flag = True
  while flag:
    flag = False
    for i in range(1,n):
      if arr[i-1] > arr[i]:
        arr[i-1],arr[i] = arr[i],arr[i-1]
        flag = True
  return arr
```

```python
bubble_sort([-5,3,2,1,-3,-3,7,2,2])
```

# 2) Insertion Sort:-

Build a sorted portion one element at a time.

Like arranging playing cards in your hand.

```python
def insertion_sort(arr):
  n  = len(arr)

  for i in range(1,n):
    for j in range(i,0,-1):
      if arr[j-1] > arr[j]:
        arr[j-1], arr[j] = arr[j], arr[j-1]
      else:
        break
  return arr

# Time : O(n**2)
# Space : O(1)
```

Stable: ✅ Yes

In-place: ✅ Yes

```python
insertion_sort([-5,3,2,1,-3,-3,7,2,2])
```

# 3). Selection Sort:-

Find the smallest element and place it at the beginning.

image.png

```python
arr = [-5,3,2,1,-3,-3,7,2,2]

def selection_sort(arr):
  n = len(arr)

    # Iterate through each position in the array
  for i in range(n):

     # Assume the current element is the smallest
    min_index = i

    # Find the actual smallest element in the remaining array
    for j in range(i+1,n):
      if arr[j] < arr[min_index]:
        min_index = j

    # Place the smallest element at its correct position
    arr[i], arr[min_index] = arr[min_index], arr[i]

  return arr
```

```python
selection_sort(arr)
```

There are two possibilities:

- A smaller element is found.
min_index is updated to its position.

- No smaller element is found.
min_index stays equal to i.

Then you perform:

arr[i], arr[min_index] = arr[min_index], arr[i]

- If min_index == i, you're simply swapping the element with itself, which has no effect.


The key idea of selection sort is:

Search first for the smallest element in the unsorted portion.
Swap once after you've finished searching.

If you swap while you're still searching, you change the array you're searching through, which can invalidate the comparisons you've already made.

# 4) Merge Sort

Divide the array into halves until each subarray has one element, then merge them back in sorted order.

image.png

```python
def merge_sort(arr):

  n = len(arr)

  if n ==1:
    return arr

  medium = len(arr)//2
  Left = arr[:medium]
  Right = arr[medium:]

  Left = merge_sort(Left)
  Right = merge_sort(Right)

  l,r = 0,0

  L_len = len(Left)
  R_len = len(Right)

  sorted_arr = [0]*n
  i=0

  while l < L_len and r < R_len:
    if Left[l] < Right[r]:
      sorted_arr[i] = Left[l]
      l+=1
    else:
      sorted_arr[i] = Right[r]
      r+=1
    i+=1

  while l < L_len:
    sorted_arr[i] = Left[l]
    l+=1
    i+=1

  while r < R_len:
    sorted_arr[i] = Right[r]
    r+=1
    i+=1


  return sorted_arr
```

```python
merge_sort([9,8,2,5,6,0,9,7,2,9])
```
