```python
# Important Concept:- Freq array

def digit_freq(arr):
  freq = [0]*10
  for i in arr:
    freq[i] +=1
  return freq
```

```python
digit_freq([3, 1, 4, 1, 2, 3,7,8,0,9])
```

```python
# Brute force method

def smallest_num(arr):

  freq = [0]*10   # Step 1: Create frequency array for digits 0–9
  for i in arr:   # Step 2: Count each digit
    freq[i] +=1

  i=0
  res = [0]*len(arr)          # Step 3: Output array of same size
  for d in range(10):         # Step 4: Traverse digits from 0 to 9
    for j in range(freq[d]):  # Step 5: Repeat each digit freq[d] times, it will skip if freq[d]=0
      res[i] = d
      i+=1
  return res
```

```python
smallest_num([6,3,4,2,7,2,1,0])
```

```python
res = smallest_number([6,3,4,2,7,2,1,0])

num = 0
for i in res:
  num = num * 10 + i
num
```

Time Complexity is O(N+k) = O(N)

The space complexity is O(n) for the output array res, which stores the sorted digits, and O(1) for the frequency array since it always has a fixed size of 10, regardless of input size. Therefore, the total space complexity is O(n).

# Count sort: This is known as Count sort. It has linear complexity, but why don't we use everywhere.

But why we have (nlogn) time complexity algorithms:

🚫 Reason 1: Counting Sort works only for integers in a small range

Counting sort needs to create a frequency array of size = max_value - min_value + 1.

👉 Example:
If your numbers are 0–9, fine — frequency array = size 10.
If your numbers are 0–1,000,000, then array = size 1 million! 😱
→ Huge memory waste → Space complexity becomes O(K), which can be very large.

# If range A[i] <= 10^6 (It works)
# If range A[i] = 10^9 (Memory Limit Exceeded error)

Online compilers allow less than 10^6 only.

🚫 Reason 2: It doesn’t work directly for non-integer or complex data

You can’t use it for:

Floating-point numbers (3.14, 2.71, …)

Strings (like “apple”, “cat”, …)

Custom objects (like Employee objects)

For such data, you need comparison-based sorting like:

Quick Sort

Merge Sort

Heap Sort
All of which have O(N log N) time complexity.

```python
def smallestNumber(A):
    MAX = 10
    n = len(A)

    # Frequency Array of size 10
    freq = [0] * MAX

    for i in range(n):
        freq[A[i]] += 1

    k = 0
    for digit in range(10):
        for i in range(1, freq[digit] + 1):
            A[k] = digit
            k += 1

    return A
```

Space Complexity - O(1)

2 Question: # Given 2 sorted integer arrays, merge them into another sorted array.

A = [2,5,8,10], B = [1,5,7,12,13]
Ans:- [1,2,5,5,7,8,10,12,13]

```python
def merge_sorted_arrays(A, B):
    n, m = len(A), len(B)
    res = [0] * (n + m)  # preallocate result array
    i = j = k = 0

    # merge step
    while i < n and j < m:
        if A[i] <= B[j]:
            res[k] = A[i]
            i += 1
        else:
            res[k] = B[j]
            j += 1
        k += 1

    # copy remaining elements from A
    while i < n:
        res[k] = A[i]
        i += 1
        k += 1

    # copy remaining elements from B
    while j < m:
        res[k] = B[j]
        j += 1
        k += 1

    return res
```

```python
A = [2,5,8,10]
B = [1,5,7,12,13]

merge_sorted_arrays(A, B)
```

```python
# Time complexity = O(n+m)
# Space complexity = O(n+m)

# Important Note:
# if res is output array then space complexity = O(N)
# if res is not output array then space complexity = O(1)
```

# Divide and conquer

```python
def merge_sort_array(arr1,arr2):
  n = len(arr1)
  m = len(arr2)
  res = [0]*(n+m)

  i=j=k=0

  while i<n and j<m:
    if arr1[i] <= arr2[j]:
      res[k]=arr1[i]
      i += 1
      k += 1
    else:
      res[k] = arr2[j]
      j += 1
      k += 1

  while i<n:
      res[k] = arr1[i]
      i += 1
      k += 1

  while j<m:
      res[k] = arr2[j]
      j += 1
      k += 1

  return res

def merge(arr):
  if len(arr) <= 1:
    return arr

  mid = len(arr)//2
  left = merge(arr[:mid])
  right = merge(arr[mid:])
  return merge_sort_array(left, right)
```

Time Complexity:
- The merge sort function recursively divides the array into halves until subarrays of size 1 are reached. This division occurs log₂(n) times, where n is the length of the input array.
- At each level of recursion, the merging process combines pairs of subarrays. Merging two sorted arrays of total size k takes O(k) time.
- Overall, since each element is involved in merging at each level, the total time across all levels sums up to O(n log n).

Space Complexity:
- The merge function creates new subarrays during each recursive call, which requires additional space proportional to the size of the array segments.
- The merge_sort_array function creates a new array res of size n + m during each merge.
- Overall, the auxiliary space used is O(n) for storing the temporary arrays during merging.
- Additionally, the recursion stack depth is O(log n), but since the main space consumption is from the temporary arrays, the total space complexity is O(n).

# 9 Qs: Given an integer array, consider first element as pivot, rearrange the elements such that for all i:
- If A[i] <= p then it should be present on left side.
- If A[i] >p then it should be present on right side.

```python
def pivot(arr):
  piv = arr[0]
  left = []
  right = []

  for i in arr[1:]:
    if i < piv:
      left.append(i)

    else:
      right.append(i)
  res = left + [piv] + right
  return res
```

```python
pivot([9, 3, 7, 1, 8, 2])
```

```python
def partition(A):
    n = len(A)
    p = A[0]           # pivot
    left = 1
    right = n - 1

    while left <= right:
        if A[left] <= p:
            left += 1
        elif A[right] > p:
            right -= 1
        else:
            A[left], A[right] = A[right], A[left]
            left += 1
            right -= 1

    # place pivot in correct position
    A[0], A[right] = A[right], A[0]

    return A   # array
```

```python
A = [7, 3, 9, 1, 8, 2, 5]
partition(A)
```

Why this algorithm works

We scan from both ends: if the left side value already belongs left, we advance left. If the right value belongs right, we decrement right. If both are misplaced, one swap corrects two items at once. When pointers cross, all elements except the pivot are partitioned; swapping pivot into right places it correctly.

Complexity

Time: O(n) — each element is examined at most once (constant-time ops).

Space: O(1) — in-place (only pointers and constants used).

# Quick Sort:-
QuickSort works using the Divide and Conquer principle:

1) Pick a pivot element.

2) Partition the array around that pivot — so all smaller (≤ pivot) elements move to the left, and larger (>) move to the right.

3) Recursively apply the same process on both sides until the array is sorted.

```python
def partition(A, low, high):
    pivot = A[low]  # first element as pivot
    left = low + 1
    right = high

    while left <= right:
        if A[left] <= pivot:
            left += 1
        elif A[right] > pivot:
            right -= 1
        else:
            A[left], A[right] = A[right], A[left]
            left += 1
            right -= 1

    A[low], A[right] = A[right], A[low]
    return right  # pivot’s final index



def quick_sort(A, low, high):
    if low >= high:
      return
    pivot_index = partition(A, low, high)
    quick_sort(A, low, pivot_index - 1)
    quick_sort(A, pivot_index + 1, high)
```

```python
A = [7, 3, 9, 1, 8, 2, 5]
print("Before Sorting:", A)
quick_sort(A, 0, len(A) - 1)
print("After Sorting:", A)
```

Time Complexity:
- Best Case: O(n log n) when the pivot divides the array into two nearly equal halves at each step, leading to balanced partitions.
- Average Case: O(n log n), assuming random or average distributions of input data, as the partitioning roughly halves the array each time.
- Worst Case: O(n^2) occurs when the pivot is consistently the smallest or largest element, resulting in highly unbalanced partitions (e.g., sorted or reverse-sorted input).

Space Complexity:
- The algorithm is in-place, so it requires O(1) additional space for swapping elements.
- However, due to recursion, the call stack depth can reach O(log n) in the best and average cases, but can degrade to O(n) in the worst case when the partitions are highly unbalanced.

In summary:
- Time complexity ranges from O(n log n) on average/best to O(n^2) in the worst case.
- Space complexity is O(log n) on average, but can be O(n) in the worst case due to recursion depth.

```python
# 2nd Method
def count_factors(n):
    cnt = 0
    i = 1
    while i * i <= n:     # loop till square root of n (no math.sqrt)
        if n % i == 0:
            if i == n // i:
                cnt += 1   # perfect square — count only once
            else:
                cnt += 2   # count both i and n//i
        i += 1
    return cnt


def sort_by_factors(arr):
    arr.sort(key=lambda x: (count_factors(x), x))
    return arr
```

Qs: Given a list of non-negative numbers, arrange them to form the largest number

```python
from functools import cmp_to_key

def compare(a,b):
  if a+b > b+a:
    return -1     # a should come before b
  elif a+b < b+a:
    return 1      # b should come before a
  else:
    return 0

def largest_number(nums):
  # convert all numbers to strings
  nums = list(map(str, nums))

  # sort using the comparator
  nums.sort(key = cmp_to_key(compare))

  # Join all numbers
  result = ''.join(nums)

  # handle case like [0, 0]
  return '0' if result[0]=='0' else result
```

```python
arr = [3, 30, 34, 5, 9]
print(largest_number(arr))
```

The function `largest_number` performs the following steps:

1. Converts all integers in the input list `nums` to strings.  
   - Time complexity: O(n), where n is the number of elements in `nums`.  
   - Space complexity: O(n), for storing the string representations.

2. Sorts the list of strings using a custom comparator `compare`.  
   - The comparison function compares two strings based on the concatenation order.  
   - Each comparison involves concatenating two strings and comparing the results.  
   - Let m be the maximum length of the string representations.  
   - Each comparison takes O(m) time due to string concatenation and comparison.  
   - The total number of comparisons in Python's sort is O(n log n).  
   - Therefore, total sorting time complexity: O(n log n * m).

3. Joins the sorted list into a single string.  
   - Time complexity: O(n * m), since concatenating all strings results in a string of length proportional to the sum of lengths.

4. Checks if the first character is '0' to handle the case of multiple zeros.  
   - Time complexity: O(1).

Overall Time Complexity:  
O(n log n * m) due to the sorting step, which dominates the other steps.

Overall Space Complexity:  
O(n * m) for storing the string representations and the final joined string.

In summary:  
Time complexity is primarily O(n log n * m), and space complexity is O(n * m).

```python
# using sorted function, which returns new res

from functools import cmp_to_key

def compare(a, b):
    if a + b > b + a:
        return -1   # a should come before b
    elif a + b < b + a:
        return 1    # b should come before a
    else:
        return 0

def largest_number(nums):
    # convert all numbers to strings
    nums = list(map(str, nums))

    # sort using custom comparator
    sorted_nums = sorted(nums, key=cmp_to_key(compare))

    # join all numbers
    result = ''.join(sorted_nums)

    # handle case like [0, 0]
    return '0' if result[0] == '0' else result


# Example-
arr = [3, 30, 34, 5, 9]
print(largest_number(arr))
```

```python
# Quick sort in Python Scaler solution

# finding the partition point & rearranging the array
def partition(array, low, high):

   # selecting the rightmost element as pivot
   pivot = array[high];

   # setting the left pointer to point at the lowest index initially
   left = low;

   #setting the left pointer to point at the lowest index initially
   right = high - 1;

   #running a loop till left is smaller than right
   while (left <= right):
       #incrementing the value of left until the value at left'th
       # index is smaller than pivot
       while (array[left] < pivot):
           left = left + 1

       #decrementing the value of right until the value at right'th
       #index is greater than pivot
       while (array[right] > pivot):
           right = right - 1;

       if (left < right):
           #swapping the elements at left & right index
           array[left], array[right] = array[right], array[left]

   #swapping pivot with the element where left and right meet
   array[left], array[high] = array[high], array[left]

   # return the partition point
   return left

# function to perform quicksort
def quickSort(array, low, high):
 if low < high:

   # since this function returns the point where the array is
   #partitioned, it is used to track the subarrays/partitions in the
   #array
   pi = partition(array, low, high)

   # recursively calling the function on left subarray
   quickSort(array, low, pi - 1)

   # recursively calling the function on right subarray
   quickSort(array, pi + 1, high)
```
