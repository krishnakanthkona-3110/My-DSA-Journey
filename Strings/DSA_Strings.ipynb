```python
#Palindrome
def is_palind(s):
  n=len(s)
  l=0
  r=n-1
  while l<r:
      if s[l]!=s[r]:
        return False
      else:
        l+=1
        r-=1
  return True
is_palind('madam')
```

```python
#Time complexity O(n) and Space Complexity O(1)
'''
Time Complexity:
- The function uses a while loop that runs until l < r.
- In each iteration, it compares characters at positions l and r.
- Since l starts at 0 and r at n-1, and both pointers move towards the center, the loop runs approximately n/2 times.
- Therefore, the time complexity is O(n), where n is the length of the string.

Space Complexity:
- The function uses a fixed amount of extra space: a few integer variables (n, l, r).
- No additional data structures are created that grow with input size.
- Therefore, the space complexity is O(1).
'''
```

```python
# Calculate the length of longest palindrome substring
```

```python
#Buteforce - For all substrings check if it is a palindrome or not.
#TC - O(n**3)
```

```python
# 1st method

def longest_palindromic_substring_bruteforce(s):
    n = len(s)
    max_len = 0

    # Step 1: Generate all substrings
    for i in range(n):
        for j in range(i, n):
            substring = s[i:j+1]

            # Step 2: Check if substring is palindrome
            if substring == substring[::-1]:
                # Step 3: Update max length
                max_len = max(max_len, len(substring))

    return max_len


# Example
print(longest_palindromic_substring_bruteforce("babad"))  # Output: 3
print(longest_palindromic_substring_bruteforce("cbbd"))   # Output: 2
```

```python
'''
Time complexity: O(n^3)
- Space complexity: O(1) (or O(n) considering temporary substring storage)
'''
```

```python
# 3nd method - without storing additional substring in a variable

def is_palind(s, i, j):
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True


def longest_palindromic_substring_bruteforce(s):
    n = len(s)
    max_len = 0

    # Step 1: Generate all substring ranges
    for i in range(n):
        for j in range(i, n):
            # Step 2: Check palindrome directly using indices
            if is_palind(s, i, j):
                max_len = max(max_len, j - i + 1)  # length = end - start + 1

    return max_len


# Example
print(longest_palindromic_substring_bruteforce("babad"))  # Output: 3 ("bab" or "aba")
print(longest_palindromic_substring_bruteforce("cbbd"))   # Output: 2 ("bb")
```

```python
#Idea - For all char and char pair consider them center and expand to find longest length.
```

```python
def longet_palind(s):
  n = len(s)
  ans = 0

  for i in range(n):
    # --- odd length palindrome ---
    l,r = i,i
    while l>=0 and r<n and s[l]==s[r]:
      ans = max(ans,r-l+1)
      l -=1
      r +=1

    # --- Even length palindrome ---
    l,r = i, i+1
    while l>=0 and r<n and s[l]==s[r]:
      ans = max(ans,r-l+1)
      l -=1
      r +=1
  return ans
```

```python
'''
The outer loop runs n times, where n is the length of s.
- For each iteration, the code performs two while loops:
  - One for checking odd-length palindromes.
  - One for checking even-length palindromes.
- In the worst case, each while loop can expand up to the entire length of the string (i.e., O(n)) in the worst case.
- Therefore, for each position, the total work is O(n) + O(n) = O(2n) = O(n).
- Combining all positions, the overall time complexity is O(n^2).

Time complexity: O(n^2)
Space complexity: O(1)
'''
```
