# Queue Types:-

1️⃣ Simple Queue – Used in task scheduling where tasks are processed in FIFO order.

2️⃣ Circular Queue – Used in CPU scheduling and buffering to reuse memory efficiently.

3️⃣ Priority Queue – Used when elements are processed based on priority instead of arrival order (e.g., OS process scheduling).

4️⃣ Deque (Double Ended Queue) – Used when insertion and deletion are required from both ends (e.g., sliding window problems).

```python
class Queue:
  def __init__(self):
    self.queue = []

  def enqueue(self, element):
    self.queue.append(element)

  def dequeue(self):
    if self.isEmpty():
      return "Queue is empty"
    return self.queue.pop(0)

  def peek(self):
    if self.isEmpty():
      return "Queue is empty"
    return self.queue[0]

  def isEmpty(self):
    return len(self.queue) == 0

  def size(self):
    return len(self.queue)
```

# Queue Implementation using Linked Lists

```python
# ==== Node ====
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None  # Pointer to the next node

class Queue:
  def __init__(self):
    self.front = None    # points to first node
    self.rear = None
    self.length = 0

  # Add element to REAR
  def enqueue(self, data):
    new_node = Node(data)

    if self.rear is None:     # queue is empty
      self.front = new_node
      self.rear = new_node
    else:
      self.rear.next = new_node  # link last node to new node
      self.rear = new_node       # move rear to new node

    self.length += 1
    print(f"Enqueued: {data}")
  # Remove element from Front
  def dequeue(self):
      if self.is_Empty():
        return "Queue is empty"

      removed = self.front.data   # save front data
      self.front = self.front.next  # move front to next node

      if self.front is None:     # if queue became empty
        self.rear = None

      self.length -= 1
      return removed

    # See front element without removing
  def peek(self):
    if self.is_Empty():
      return "Queue is empty"
    return self.front.data

      # Check if empty
  def is_Empty(self):
    return self.length ==0

    # size of queue
  def size(self):
    return self.length

  # display queue
  def display(self):
    current = self.front

    while current:
      print(current.data, end="->")
      current = current.next
    print()

```

```python
q = Queue()

# Enqueue
q.enqueue("A")
q.enqueue("B")
q.enqueue("C")
q.enqueue("D")

q.display()

print(q.peek())
print(q.dequeue())
print(q.size())
print(q.is_Empty())
```

# Deque (Double Ended Queue) –

Deque is a linear data structure where insertion and deletion are allowed at both ends.

```python
from collections import deque

dq = deque()

# insert at rear
dq.append(10)
dq.append(20)

# insert at front
dq.appendleft(5)

print(dq)
```

```python
from collections import deque

class Queue:
    def __init__(self):
        self.queue = deque()          # faster than list!

    def enqueue(self, item):
        self.queue.append(item)       # add to rear

    def dequeue(self):
        if self.is_empty():
            return "Queue is Empty!"
        return self.queue.popleft()   # O(1) removal from front!

    def peek(self):
        return self.queue[0] if self.queue else "Empty"

    def is_empty(self):
        return len(self.queue) == 0

    def size(self):
        return len(self.queue)

    def display(self):
        print(f"Queue (front→rear): {list(self.queue)}")


# Usage
q = Queue()
q.enqueue("A")
q.enqueue("B")
q.enqueue("C")
q.display()         # Queue (front→rear): ['A', 'B', 'C']
print(q.dequeue())  # A
print(q.dequeue())  # B
q.display()         # Queue (front→rear): ['C']
```

# Circular Queue - A Circular Queue connects the last position back to the first to reuse empty spaces.

- If front moves forward, new elements can reuse earlier spaces.


CPU scheduling

Streaming buffers

Multiplayer game loops

```python
class CircularQueue:
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = self.rear = -1

    def enqueue(self, item):
        if (self.rear + 1) % self.size == self.front:
            print("Queue is Full!")
            return
        elif self.front == -1:        # first element
            self.front = self.rear = 0
        else:
            self.rear = (self.rear + 1) % self.size
        self.queue[self.rear] = item
        print(f"Enqueued: {item}")

    def dequeue(self):
        if self.front == -1:
            return "Queue is Empty!"
        item = self.queue[self.front]
        if self.front == self.rear:   # last element
            self.front = self.rear = -1
        else:
            self.front = (self.front + 1) % self.size
        return item

    def display(self):
        print(f"Queue: {self.queue}")


cq = CircularQueue(5)
cq.enqueue(1)
cq.enqueue(2)
cq.enqueue(3)
print(cq.dequeue())  # 1
cq.enqueue(4)        # reuses position 0!
cq.display()
```

# Priority Queue - In a Priority Queue, elements are removed based on priority, not FIFO

- Python provides a built-in implementation using heapq.


Dijkstra shortest path

Task scheduling

Network routing

```python
import heapq

class PriorityQueue:
    def __init__(self):
        self.queue = []

    def enqueue(self, item, priority):
        heapq.heappush(self.queue, (priority, item))

    def dequeue(self):
        if self.queue:
            return heapq.heappop(self.queue)[1]  # returns highest priority item
        return "Empty!"

    def display(self):
        print(f"Queue: {self.queue}")


pq = PriorityQueue()
pq.enqueue("low task", 3)
pq.enqueue("urgent task", 1)     # priority 1 = highest
pq.enqueue("medium task", 2)

print(pq.dequeue())  # urgent task  ← comes out first!
print(pq.dequeue())  # medium task
print(pq.dequeue())  # low task
```

# Implement Queue using Two Stacks
Use two stacks:

stack1 → for enqueue

stack2 → for dequeue

```python
class MyQueue:

    def __init__(self):
        self.s1 = []
        self.s2 = []

    def enqueue(self, x):
        self.s1.append(x)

    def dequeue(self):

        if not self.s2:
            while self.s1:
                self.s2.append(self.s1.pop())

        if not self.s2:
            return "Queue Empty"

        return self.s2.pop()
```

```python
q = MyQueue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.dequeue())   # 10
```

# BFS using Queue

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        node = queue.popleft()        # dequeue
        print(node, end=" → ")

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor) # enqueue

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [], 'E': [], 'F': []
}

bfs(graph, 'A')
```
