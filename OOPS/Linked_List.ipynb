# Linked List
A linked list consists of nodes with some sort of data, and a pointer, or link, to the next node.

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None
```

```python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)
n4 = Node(40)
```

```python
head = n1
n1.next = n2
n2.next = n3
n3.next = n4
n4.next = None
```

```python
temp = head

while temp != None:
  print(temp.data)
  temp = temp.next
```

To reaach 3rd node = 2 jumps

To reach kth node = (k-1) jumps
