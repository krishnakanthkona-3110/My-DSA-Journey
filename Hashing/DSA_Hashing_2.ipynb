# Pigeonhole Principle

Statement:
If you put n items (pigeons) into m containers (holes) and n > m, then at least one container must have more than one item.

✅ Example

5 holes

6 pigeons

Since 6 > 5, collision will always happen.
At least one hole will contain ≥ 2 pigeons.

Relation to Hashing

In hashing:

Keys / values → pigeons

Hash table slots → holes

If:

Number of keys > number of slots
➡️ Collision is guaranteed (by Pigeonhole Principle).

🔹 What is a Collision?

A collision occurs when:

Two different keys produce the same hash index

🔹 Can Collisions be Eliminated?

❌ No, collisions cannot be completely avoided
✔️ They can only be minimized and handled efficiently


# 🔹 Collision Resolution Techniques

1) Open Hashing (Chaining)

Open hashing is a collision resolution technique in hashing where each index of the hash table stores a collection (usually a linked list) of elements.

⚠️ Note (important for interviews):
Open hashing = Chaining
Closed hashing = Open addressing

Many people confuse the terms.

Definition

In open hashing, when two or more keys hash to the same index:

They are stored together at that index

Using a linked list / dynamic structure

🔹 Example

Hash table size = 5
Keys = 12, 22, 32

Hash Table

Index 2 → 12 → 22 → 32

🔹 Why Open Hashing is Needed

By the Pigeonhole Principle:

If number of keys > number of slots

Collisions are unavoidable

Open hashing allows multiple keys per slot

🔹 Advantages

✔️ Simple implementation
✔️ No limit on number of elements
✔️ Less sensitive to table size
✔️ Easy deletion

🔹 Disadvantages

❌ Extra memory for pointers
❌ Performance degrades if chains become long
❌ Worst-case search becomes O(n)

# Longest Consecutive Sequence

Q1. Given an unsorted integer array A of size N.


Find the length of the longest set of consecutive elements from array A.

Example Input

Input:

A = [100, 4, 200, 1, 3, 2]


Example Output

Output:

 4

# Brute Force

```python
def longest_consecutive_bruteforce(arr):

  n = len(arr)
  longest = 0

  for i in range(n):
    current = arr[i]
    count = 1

    while (current+1) in arr:
      current += 1
      count += 1

    longest = max(longest, count)

  return longest

```

```python
longest_consecutive_bruteforce([1,2,10,9,4,3])
```

# Time and space complexity
The time complexity of the given function is O(n^2). This is because for each element in the array (outer loop), it potentially performs a search for the next consecutive number using the expression `(current+1) in arr`, which is an O(n) operation since `arr` is a list and the `in` operator performs a linear search. Therefore, in the worst case, for each of the n elements, you might perform up to n searches, leading to O(n^2).

The space complexity is O(1), as the function uses only a fixed amount of additional space regardless of the input size. It maintains a few variables (`n`, `longest`, `current`, `count`) but does not allocate extra space proportional to the input size.

Key observation ⚠️

The expensive operation here is:

(current + 1) in arr


➡️ arr is a list, not a set.

Checking membership in a list takes O(n) time in the worst case.

# Optimized Solution

```python
def longest_consecutive_optimized(arr):
  hm = {}

  # step-1: put all elements as True
  for x in arr:
    hm[x] = True

  # step-2: mark non-starting elements as False
  for x in arr:
    if x-1 in hm:
      hm[x] = False

  # step-3: find max length
  max_length = 0

  for x in arr:
    if hm[x] is True:
      length =0
      curr = x

      while curr in hm:
        length += 1
        curr += 1

      max_length = max(max_length, length)

  return max_length
```

```python
longest_consecutive_optimized([11,10,1,2,4,3,9])
```

Time Complexity:
- The initial loop that inserts all elements into the hash map runs in O(n), where n is the number of elements in the array.
- The second loop that marks non-starting elements as False also runs in O(n). Each lookup in the hash map is O(1) on average.
- The third loop iterates over all elements, and for each starting element (where hm[x] is True), it counts the length of the consecutive sequence. In the worst case, each element is part of exactly one sequence, and the inner while loop advances through the sequence without revisiting elements. Overall, each element is visited at most once during these sequence expansions.
- Combining these, the total time complexity is O(n).

Space Complexity:
- The hash map hm stores all elements, so it uses O(n) space.
- Additional variables use constant space.
- Therefore, the overall space complexity is O(n).

In summary:
Time complexity: O(n)
Space complexity: O(n)

```python
# 3rd approach

def longest_consecutive(A):
    if not A:
        return 0

    num_set = set(A)
    longest = 0

    for num in num_set:
        # check if num is start of a sequence
        if num - 1 not in num_set:
            current = num
            length = 1

            while current + 1 in num_set:
                current += 1
                length += 1

            longest = max(longest, length)

    return longest
```

#  Longest Subarray Zero Sum
Q2. Given an array A of N integers.
Find the length of the longest subarray in the array which sums to zero.

If there is no subarray which sums to zero then return 0.

```python
#Brute Force

def longest_zero_sum_subarray_bruteforce(A):
    n = len(A)
    max_len = 0

    for i in range(n):
        for j in range(i, n):
            curr_sum = 0
            for k in range(i, j + 1):
                curr_sum += A[k]

            if curr_sum == 0:
                max_len = max(max_len, j - i + 1)

    return max_len
```

```python
longest_zero_sum_subarray_bruteforce([1, -1, 3, 2, -2, -3])
```

```python
# Time complexity: O(n^3)
# Space complexity: O(1)
```

# Optimal Approach (O(n) Time, O(n) Space)

```python
def longest_zero_sum(arr):
  n = len(arr)
  prefix_sum = 0
  max_len = 0
  hm = {}

  for i in range(n):
    prefix_sum += arr[i]

    #case 1: subarray from 0 to i
    if prefix_sum == 0:
      max_len = i+1

    # case 2: prefix_sum seen before
    if prefix_sum in hm:
      max_len = max(max_len, i-hm[prefix_sum])
    else:
      hm[prefix_sum] = i

  print(hm)

  return max_len
```

#  Sort Array in given Order

3Q. Given two arrays of integers A and B, Sort A in such a way that the relative order among the elements will be the same as those are in B.
For the elements not present in B, append them at last in sorted order.

Return the array A after sorting from the above method.

NOTE: Elements of B are unique.

Input 1:

A = [1, 2, 3, 4, 5, 4]
B = [5, 4, 2]

Output 1:

[5, 4, 4, 2, 1, 3]



```python
def relative_sort(A,B):
  freq = {}

  # step-1: frequency of A
  for x in A:
    freq[x]=freq.get(x,0) +1

  result = []

  # step-2: elements present in B (in B's order)
  for x in B:
    if x in freq:
      result.extend([x]*freq[x])
      del freq[x]

  # step-3: remaining elements (sorted)
  remaining = sorted(freq.keys())

  for x in remaining:
    result.extend([x]*freq[x])

  return result
```

```python
A = [1, 2, 3, 4, 5, 4]
B = [5, 4, 2]
relative_sort(A,B)
```

Time Complexity:

1. Building the frequency dictionary (freq):
   - Iterates through all elements in A once.
   - Time complexity: O(n), where n is the length of A.

2. Processing elements in B:
   - For each element in B, checking if it exists in freq and updating result.
   - Each lookup and deletion in a dictionary is O(1) on average.
   - Total for this step: O(m), where m is the length of B.

3. Sorting remaining keys:
   - The remaining keys are the elements in freq not in B.
   - Let k be the number of remaining unique elements.
   - Sorting these keys takes O(k log k).

4. Extending result with remaining elements:
   - Extending the result list based on the frequency counts.
   - Total time proportional to the total number of remaining elements, which is less than or equal to n.

Overall, the dominant factors are building the frequency dictionary (O(n)), processing B (O(m)), and sorting remaining keys (O(k log k)). Since k ≤ n, the total time complexity is approximately:

O(n + m + k log k) ≤ O(n log n) in the worst case.

Space Complexity:

- The frequency dictionary (freq) stores counts for each unique element in A.
- Space used: O(n) in the worst case (all elements are unique).
- The result list grows to size n.
- Additional space for the remaining keys and temporary lists is negligible compared to the main data structures.

Overall, space complexity is O(n).

# 4Q. Count Subarrays

Misha likes finding all Subarrays of an Array. Now she gives you an array A of N elements and told you to find the number of subarrays of A, that have unique elements.

Since the number of subarrays could be large, return value % 109 +7.



```python
# Brute Force method : O(n^3)

def count_unique_subarrays_bruteforce(A):

  n = len(A)
  count = 0

  for i in range(n):
      for j in range(i, n):
          seen = set()
          unique = True

          for k in range(i, j + 1):
              if A[k] in seen:
                  unique = False
                  break
              seen.add(A[k])

          if unique:
              count += 1

  return count
```

```python
count_unique_subarrays_bruteforce([1,2,3,2])
```

```python
# above method with comments, for better understanding...

def brute_force(A):

    # Length of the input array
    n = len(A)

    # Stores the total number of valid subarrays
    # that contain only unique elements
    count = 0

    # Choose every index as the starting point
    # of a subarray
    for i in range(n):

        # Fresh set for each starting index
        # Used to track elements already seen
        # in the current subarray
        s = set()

        # Extend the subarray from i to n-1
        for j in range(i, n):

            # If current element is already present,
            # the subarray is no longer unique
            if A[j] in s:

                # No need to continue further because
                # any larger subarray will also contain
                # this duplicate element
                break

            # Add current element to the set
            s.add(A[j])

            # Current subarray A[i:j+1] contains
            # all unique elements, so count it
            count += 1

    # Return total number of unique-element subarrays
    return count
```

```python
brute_force([1,2,3,4])
```

```python
# Optimized solution

def count_unique_subarrays(A):
    n = len(A)
    seen = set()
    left = 0
    ans = 0

    for right in range(n):
        while A[right] in seen:
            seen.remove(A[left])
            left += 1

        seen.add(A[right])
        ans += right - left + 1

    return ans
```

```python
count_unique_subarrays([1,2,3,4])
```

```python
# above solution with comments

def count_unique_subarras(A):

    # Length of the input array
    n = len(A)

    # Stores elements currently present in the sliding window
    # Helps us quickly check for duplicates in O(1) time
    seen = set()

    # Left pointer of the sliding window
    left = 0

    # Stores total count of valid subarrays
    ans = 0

    # Expand the window by moving right pointer
    for right in range(n):

        # If current element already exists in the window,
        # shrink the window from the left until the duplicate is removed
        while A[right] in seen:

            # Remove leftmost element from the window
            seen.remove(A[left])

            # Move left pointer forward
            left += 1

        # Add current element to the window
        # Now the window contains only unique elements
        seen.add(A[right])

        # Number of valid subarrays ending at 'right'
        #
        # Current valid window:
        # [left .... right]
        #
        # All these subarrays are valid:
        # [A[right]]
        # [A[right-1], A[right]]
        # ...
        # [A[left], ..., A[right]]
        #
        # Count = window size
        ans += (right - left + 1)

    return ans
```

Maintain a sliding window containing only unique elements. For every position right, the number of unique-element subarrays ending at right equals the current window size (right - left + 1), which is added to the answer. This gives an O(n) solution.

```python
count_unique_subarras([1,2,3,4,1])
```

This uses the sliding window technique to count all subarrays with distinct elements. We maintain a window [left, right] containing unique elements using a set. If a duplicate is found, we shrink the window from the left until the duplicate is removed. For every valid window, the number of unique-element subarrays ending at right is right - left + 1, which we add to the answer. Since each element is added and removed from the set at most once, the overall time complexity is O(n) and space complexity is O(n).

Time Complexity	O(n)

Space Complexity O(n)

# 5Q. Colorful Number
Given a number A, find if it is COLORFUL number or not.

If number A is a COLORFUL number return 1 else, return 0.

What is a COLORFUL Number:

A number can be broken into different consecutive sequence of digits.
The number 3245 can be broken into sequences like 3, 2, 4, 5, 32, 24, 45, 324, 245 and 3245.
This number is a COLORFUL number, since the product of every consecutive sequence of digits is different

```python
def colorful_bruteforce(A):
    digits = list(map(int, str(A)))
    n = len(digits)
    products = set()

    for i in range(n):
        for j in range(i, n):
            prod = 1
            for k in range(i, j + 1):
                prod *= digits[k]

            if prod in products:
                return 0
            products.add(prod)

    return 1
```

```python
colorful_bruteforce([1,2,3])
```

Time complexity:
- Generating all subarrays: O(n^2).
- Computing the product for each subarray: in the worst case, O(n) per subarray.
- Total worst-case time: O(n^3).

Space complexity:
- The set 'products' stores at most all distinct products of subarrays. In the worst case, the number of subarrays is O(n^2), so the set can hold up to O(n^2) products.
- Additional space for the digits list: O(n).

Overall:
- Time complexity: O(n^3).
- Space complexity: O(n^2).
