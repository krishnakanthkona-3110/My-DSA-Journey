```python
'''
N = 6 [1 2 3 1 2 5] -- Ans =3
N = 8 [4 3 3 2 5 6 4 5] -- Ans = 2
N = 7 [2 6 8 4 7 2 9] -- Ans = 6
'''
```

```python
# Brute Force - We can use two loops:

# Outer loop to pick each element.

# Inner loop to check if it repeats anywhere else.
```

```python
def first_repeat(arr):
  n = len(arr)

  for i in range(n):
    for j in range(n):
      if i!=j and arr[i]==arr[j]:
        break # found duplicate, go to next i
    else:
          # This else belongs to the for-loop, not the if
          # It runs only if 'break' didn't happen
      return arr[i]
  return None
```

```python
first_repeat([6,2,2,3,3,5,1,1,2,5])
```

```python
'''Time: O(N²)

Space: O(1)'''
```

```python
# Using flag

def first_non_repeating(arr):
    n = len(arr)
    for i in range(n):
        unique = True
        for j in range(n):
            if i != j and arr[i] == arr[j]:
                unique = False
                break
        if unique:
            return arr[i]
    return None
```

```python
first_non_repeating([2,2,3,3,1,1,2,5])
```

```python
# Optimzed solution.
# a) Store freq of all elements.
# b) First element with freq=1 is answer.

# Step 1: create an empty dictionary
# Step 2: count occurrences of each element
# Step 3: find the first element with count == 1
```

```python
def first_non_repeat(arr):

  freq = {}

  for i in arr:
    freq[i]=freq.get(i,0)+1

  for i in freq:
    if freq[i]==1:
      return i

  return None
```

Optimzed solution

```python
def pairs(arr,k):

  freq = set() # to store elements we've already visited
  n = len(arr)

  for i in range(n):
    if (k-arr[i]) in freq:
      return True, freq
    else:
      freq.add(arr[i]) # store current element for future checks

  return False, freq
```

```python
pairs([8,9,1,-2,4,5,11,-6,4],6)
```

```python
# - Time complexity: O(n)
# - Space complexity: O(n)
```

# 6th Qn: Count the number of pairs with sum = k

```python
# brute force method

def count_pair(arr,k):
  ans = 0
  n = len(arr)

  for i in range(n):
    for j in range(i+1,n):
      if arr[i]+arr[j]==k:
        ans +=1
  return ans
```

```python
count_pair([8,9,1,-2,4,5,11,-6,4],6)
```

```python
# Time complexity: O(n^2)
# Space complexity: O(1)
```

We iterate from left → right. For each x we only count complements that appeared earlier (stored in hm). That ensures each unordered pair (i, j) with i < j is counted exactly once when we reach the later index j.

Because we add hm[x] after checking complements, the current element does not pair with itself or with elements that come later — only with earlier ones already in hm.

```python
# 2nd method

def count_pairs(arr,k):
  ans = 0
  hm = {}
  n = len(arr)
  for i in range(n):
    if k-arr[i] in hm:
      ans += hm[k-arr[i]]            # add how many times complement appeared
    hm[arr[i]]= hm.get(arr[i],0) +1  # store/update frequency

  return ans
```

```python
count_pairs([8,9,1,-2,4,5,11,-6,4],6)
```

```python
count_pairs([2,5,2,5,8,5,2,8],10)
```

```python
# Time complexity: O(n)
# Space complexity: O(n)

# In the worst case, all elements are unique, so the hash map will contain n entries.
```

## 7 Qs: Given an array of N elements. check if there exists a subarray with sum equal to 0.

Subarray:- A subarray is a contiguous (continuous) part of an array.
That means — it must include consecutive elements (no skipping).

```python
# Brute Force:

# Generate all possible subarrays using 2 loops (i and j).
# For each subarray, calculate its sum with another loop (k).
# If the sum is 0 → return True.
```

```python
def brute_sum_sub(arr):
  n = len(arr)
  for i in range(n):          # starting index
    for j in range(i,n):      # ending index
      total = 0
      for k in range(i,j+1):
        total += arr[k]        # sum of subarray arr[i...j]

      if total == 0:
        return True
  return False
```

```python
brute_sum_sub([4, 2, -3, 1, 6])
```

1. The outer loop runs from i = 0 to n-1, so it executes n times.
2. The middle loop runs from j = i to n-1, which on average runs about n/2 times, but in the worst case, it can be up to n times.
3. The innermost loop runs from k = i to j, which in the worst case can be up to n times.

Overall, the time complexity is O(n^3).

Regarding space complexity, the function uses a constant amount of extra space: variables i, j, k, total, and n. No additional data structures proportional to n are used.

Thus, the space complexity is O(1).

```python
def brute2_sum(arr):
  n = len(arr)

# Outer loop picks the starting index of subarray
  for i in range(n):
    sum = 0
    # Inner loop expands the subarray from index i to end
    for j in range(i,n):
      sum += arr[j]
      if sum == 0:
        return True
  return False
```

Outer loop chooses the start index i

Inner loop extends subarray to j

Maintain running sum instead of recomputing

If sum == 0 → subarray found → return True

```python
# The time complexity is O(n^2)
# Space complexity is O(1)
```

Optimzed Solution:
Using Prefix Sum + Hash Set, O(N))

```python
def prefix_sum(arr):
  prefix_sum = 0
  n = len(arr)
  seen = set()

  for i in range(n):
    prefix_sum += arr[i]
    if prefix_sum == 0 or prefix_sum in seen:
      return True
    seen.add(prefix_sum)
  return False
```

```python
prefix_sum([4, 2, -3, 1, 6])
```

If at any point:

prefix_sum == 0 → means the subarray from start (0) to that index has sum = 0.

If same prefix_sum repeats, then the elements between those two positions sum to 0.



```python
# Time complexity: O(n), where n is the length of the array, because it processes each element exactly once.

# Space complexity: O(n), in the worst case, the set 'seen' could store all prefix sums, which is proportional to the size of the array.
```

# 8 Qs: Check if subarray sum=k

```python
def subarray_sum(arr, k):
    prefix_sum = 0          # Running cumulative sum
    seen = set()            # Stores previous prefix sums

    for i in range(len(arr)):

        prefix_sum += arr[i]   # Add current element

        # Case 1: If prefix sum itself equals k
        if prefix_sum == k:
            return True

        # Case 2: If (prefix_sum - k) seen before
        # That means subarray exists with sum k
        if (prefix_sum - k) in seen:
            return True

        # Store current prefix sum
        seen.add(prefix_sum)

    return False
```

```python
subarray_sum([4, 2, -3, 1, 6],6)
```

```python
# Time complexity: O(n)
# Space complexity: O(n)
```

```python
def pre_sum(arr,k):
  pre_sum = 0
  n = len(arr)
  seen = set()

  for i in range(n):
    pre_sum += arr[i]

    if pre_sum == k or (pre_sum-k) in seen:
      return True

    seen.add(pre_sum)
  return False
```

```python
pre_sum([4, 2, -3, 1, 6],6)
```
