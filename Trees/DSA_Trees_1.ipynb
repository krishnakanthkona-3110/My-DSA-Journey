# Iterative In-order Traversal:

Why Iterative Inorder is better than Recursive Inorder

👉 Because it is safer for big trees.

🌳 Recursive Inorder (normal)

Uses recursion

Recursion uses the system call stack

If the tree is very deep → stack overflow error

❌ Risky for large inputs

🌳 Iterative Inorder

Uses our own stack

No recursion

Works even for very deep trees

✅ Safe and reliable


# “Iterative inorder is better because it avoids recursion stack overflow.”

        1
       / \
      2   3
     / \
    4   5


```python
class TreeNode:
  def __init__(self,data):
    self.data = data
    self.left = None
    self.right = None

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
```

```python
def inorder_iterative(root):
  curr = root
  st = []

  while curr is not None or len(st) >0:
    if curr is not None:
      st.append(curr)
      curr = curr.left

    else:
      curr = st.pop()
      print(curr.data)
      curr = curr.right
```

```python
inorder_iterative(root)
```

# Q2: Problem Description

Given a binary tree, return the inorder traversal of its nodes' values.

Output Format

Return an integer array denoting the inorder traversal of the given binary tree.

```python
def inorder_iterative_2(root):
  res = []
  stack = []
  curr = root

  while curr is not None or len(stack) > 0:
    if curr is not None:
      stack.append(curr)
      curr = curr.left

    else:
      curr = stack.pop()
      res.append(curr.data)
      curr = curr.right

  return res
```

```python
inorder_iterative_2(root)
```

# Both recursive and iterative inorder traversals take O(N) time, but iterative avoids recursion stack overflow

# Iterative Pre-order Traversal:

Iterative preorder uses a stack, prints the node when popped, and pushes right before left

```python
def preorder_iterative(root):
  if root is None:
    return

  stack = [root]

  while len(stack) > 0:
    node = stack.pop()
    print(node.data)

    # Push right first so left is processed first
    if node.right is not None:
      stack.append(node.right)
    if node.left is not None:
      stack.append(node.left)
```

```python
preorder_iterative(root)
```

# Iterative Post-order Traversal:

```python
# 2 Stacks – Beginner Friendly

def postorder_iterative(root):
  if root is None:
    return

  stack1 = [root]
  stack2 = []

  while len(stack1) > 0:
    node = stack1.pop()
    stack2.append(node)

    if node.left is not None:
      stack1.append(node.left)
    if node.right is not None:
      stack1.append(node.right)

  while len(stack2) > 0:
    node = stack2.pop()
    print(node.data)

```

```python
postorder_iterative(root)
```

# --- Use two stacks. The first stack performs a modified preorder traversal, and the second stack reverses it to get postorder.

# LEVEL ORDER TRAVERSAL:-
Level Order Traversal (Breadth First Search) of Binary Tree


Visit nodes level by level, left to right

```python
# Level Order Traversal (Using List as Queue)

def level_order_list(root):
  if root is None:
    return

  queue = []
  queue.append(root)     # # enqueue

  while len(queue) >0:
    node = queue.pop(0)   # # dequeue
    print(node.data)

    if node.left is not None:
      queue.append(node.left)
    if node.right is not None:
      queue.append(node.right)
```

```python
level_order_list(root)
```

```python
# 2D using List

def levelorder_list_2d(root):
    if root is None:
        return []

    result = []
    queue = []
    queue.append(root)

    while len(queue) > 0:
        level = []
        size = len(queue)   # number of nodes in current level

        for _ in range(size):
            node = queue.pop(0)
            level.append(node.data)

            if node.left is not None:
                queue.append(node.left)
            if node.right is not None:
                queue.append(node.right)

        result.append(level)

    return result
```

```python
levelorder_list_2d(root)
```

Time complexity is O(N²) because list pop(0) is O(N).
Space complexity is O(N) due to the queue.

The time complexity of the given level order traversal using a list as a queue is O(n), where n is the number of nodes in the binary tree. This is because each node is enqueued and dequeued exactly once. However, since list.pop(0) has a time complexity of O(n) (due to shifting elements), the overall time complexity can degrade to O(n^2) in the worst case for large trees.

The space complexity is O(n), as the queue can hold up to all nodes at the last level of the tree in the worst case, which is proportional to the number of nodes in the tree. Additionally, the space used by the tree itself is O(n).

# Using queue

```python
from collections import deque

def level_order_queue(root):
  if root is None:
    return

  queue = deque()
  queue.append(root)

  while len(queue) > 0:
    node = queue.popleft()
    print(node.data)

    if node.left is not None:
      queue.append(node.left)
    if node.right is not None:
      queue.append(node.right)
```

```python
level_order_queue(root)
```

The time complexity of the level_order_queue function is O(n), where n is the number of nodes in the binary tree. This is because each node is enqueued exactly once and dequeued exactly once, leading to a linear traversal of all nodes.

The space complexity is also O(n) in the worst case. This occurs when the tree is very broad, such as a complete binary tree at the last level, where the queue may hold up to all nodes at that level simultaneously. Therefore, the maximum size of the queue is proportional to the number of nodes at the widest level of the tree.

# Q1:Level Order:-  Given a binary tree, return the level order traversal of its nodes' values. (i.e., from left to right, level by level).

Output Format

Return a 2D integer array denoting the level order traversal of the given binary tree.

```python
from collections import deque

def levelorder_queue_2d(root):
  if root is None:
    return []

  result = []
  queue = deque()
  queue.append(root)

  while len(queue) >0:
    level = []
    size = len(queue) # number of nodes in current level

    for _ in range(size):
      node = queue.popleft()
      level.append(node.data)

      if node.left is not None:
        queue.append(node.left)
      if node.right is not None:
        queue.append(node.right)

    result.append(level)

  return result
```
