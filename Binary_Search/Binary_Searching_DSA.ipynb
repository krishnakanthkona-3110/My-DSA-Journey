# What is Binary Search?

Binary Search is an efficient algorithm to find an element’s position in a sorted list.

It works on divide and conquer principle.

Instead of checking elements one by one (like linear search), it repeatedly divides the search space in half.

## To apply Binary Search:-

1) Think of Target,
2) Search - space, and
3) condition on the basis of which search space can be reduced by half

Q1: Given a sorted arr[]. Search an element k. If k is present return it's index. Otherwise return -1.

```python
# brute force - Linear Search, check each element one by one.

def search(arr,k):
  n = len(arr)
  for i in range(n):
    if arr[i]==k:
      return i
  return -1
```

```python
arr = [2, 4, 6, 8, 10, 12]
search(arr,11)
```

```python
# Time complexity - O(n)
# Space complexity - O(1)
```

```python
# Binary search

def binary_search(arr,k):
  low = 0
  high = len(arr)-1

  while low <= high:
    mid = (low+high)//2   # find middle index

    if arr[mid]==k:
      return mid          # element found
    elif arr[mid]<k:
      low = mid +1        # search right half
    else:
      high = mid -1       # search left half

  return -1
```

```python
binary_search([2, 4, 6, 8, 10, 12], 80)
```

# 2Q: Given a sorted arr[N]. Find first occurence of k.

Target - find first index of k.

Search space - 0 to (n-1)

```python
# Brute force - linear search, check each element one by one
```

```python
# Binary search

def first_occurence(arr,k):
  n = len(arr)
  low = 0
  high = n-1
  result = -1

  while low <= high:
    mid = (low+high)//2

    if arr[mid]==k:
      result = mid        # found, but keep searching left array
      high = mid -1
    elif arr[mid] < k:
      low = mid + 1
    else:
      high = mid -1

  return result
```

```python
arr = [1, 2, 2, 2, 3, 4, 5]
k = 2
first_occurence(arr, k)
```

Given a sorted arr[N]. Find last occurence of k.

```python
def last_occurence(arr,k):
  n = len(arr)
  low = 0
  high = n-1
  res = -1

  while low <= high:
    mid = (low+high)//2

    if arr[mid]==k:
      res = mid         # found → store index
      low =mid +1       # keep searching on the right
    elif arr[mid] < k:
      low = mid + 1
    else:
      high = mid - 1

  return res
```

```python
arr = [1, 2, 2, 2, 3, 4, 5]
k = 2

last_occurence(arr, k)
```

```python
def first_local_minima(arr):
  n = len(arr)

  if n==1:          # Handle single element array
    return 0

  if arr[0] <= arr[1]:      # Check first element
    return 0

  for i in range(1,n-1):
    # Check middle elements
    if arr[i] < arr[i-1] and arr[i] < arr[i+1]:
      return i

  if arr[n-1] <= arr[n-2]:      # check last element
    return (n-1)

  return -1
```

Can we apply B.S only on sorted array?

Non sorted arrays can also have BS.

```python
def lcl_minima(arr):
    n = len(arr)

    # Edge cases
    if n == 0:
        return -1

    if n == 1:
        return 0

    if arr[0] < arr[1]:
        return 0

    if arr[n - 1] < arr[n - 2]:
        return n - 1

    low = 1
    high = n - 2

    while low <= high:
        mid = (low + high) // 2

        # Found local minimum
        if arr[mid] < arr[mid - 1] and arr[mid] < arr[mid + 1]:
            return mid

        # Going downhill towards the left
        elif arr[mid] > arr[mid - 1]:
            high = mid - 1

        # Going downhill towards the right
        else:
            low = mid + 1

    return -1
```

# 4Q. Find the square root of a number. Integral part


```python
# 1st Method: Brute Force Approach:

def sqroot(n):
  i = 1
  while i*i <=n:
    i += 1
  return i-1
```

```python
sqroot(100)
```

```python
# Time Complexity - O(n)
# Space Complexity - O(1)
```

```python
# 2nd Method - Binary Search

def sqrootbin(n):
   low = 0
   high= n
   ans = -1

   while low <= high:
    mid = (low+high)//2

    if mid*mid == n:
      return mid
    elif mid*mid < n:
      ans = mid
      low = mid +1
    else:
      high = mid -1

   return ans
```

```python
sqrootbin(101)
```

```python
# Time complexity: O(log n)
# Space complexity: O(1)
```

# Qs 5: Painter's Partition Problem

Given N boards with length of each board.

(a) A painter takes 1 unit of time to paint 1 unit of length.

(b) A board can only be painted by 1 painter.

(c) A painter can only paint boards placed next to each other (i.e continuous segment)

Qs: Find minimum time to paint all boards if P painters are available.

Target: min time to paint all boards, with k workers.

Search Space: min time is max(all board lengths), --- max time (1 worker - sum of length of all boards)

```python
# Helper Function
'''arr - list of board lengths
  n - number of boards
  p - number of painters available
  mid - max time a painter is allowed to paint
'''

def canpaint(arr, n, p, mid):
  painters = 1    # Start with 1 painter
  curr_sum = 0    # Time used by current painter

  for length in arr:
    if curr_sum + length <= mid:      # Case 1: current painter can take this board
      curr_sum += length

    else:
      painters += 1                     # Need a new painter
      curr_sum = length

      if painters > p:                  # Case 2: if painters become more than P -> not possible
        return False

  return True                 # If loop ends without exceeding painters → possible
```

```python
def partition_painters(arr,p):
  n = len(arr)

  low = max(arr)         # min possible time
  high = sum(arr)        # max possible time

  ans = high

  while low <= high:
    mid = (low+high)//2

    if canpaint(arr,n,p,mid):
      ans = mid
      high = mid - 1        # try for better (smaller time)

    else:
      low = mid + 1       # need more time

  return ans
```

```python
arr = [10, 20, 30, 40]
p = 2
partition_painters(arr,p)
```

```python
# Time complexity: O(n * log(sum(arr) - max(arr)))
# Space complexity: O(1)
```

```python
def search_rotated(arr,k):
  n = len(arr)
  low = 0
  high = n-1

  while low<= high:
    mid = (low+high)//2

  # Element found
    if arr[mid] == k:
      return mid

  # Left array is sorted
    if arr[low] <= arr[mid]:
      if arr[low] <= k < arr[mid]:
        high = mid - 1      # search in left half
      else:
        low = mid + 1       # search in right half

    # RIGHT HALF is sorted
    else:
      if arr[mid] < k <= arr[high]: # search in right half
        low = mid + 1
      else:
        high = mid - 1               # search in left half

  return -1
```

```python
arr = [4,5,6,7,1,2,3]
k = 2
search_rotated(arr, k)
```

Time Complexity:
- The algorithm divides the search space approximately in half during each iteration of the while loop.
- This halving process continues until the element is found or the search space is exhausted.
- Therefore, the time complexity is O(log n), where n is the length of the array.

# 7 Qs. Median of two sorted arrays.


```python
# Brute Force - Merge two sorted arrays, ruturn middle one
# Time Complexity - O(m+n)
```

```python
def findMedianSortedArrays_simple(nums1, nums2):
    m, n = len(nums1), len(nums2)
    total = m + n
    mid = total // 2

    i = j = 0
    prev = curr = 0

    # Traverse until we reach the middle
    for _ in range(mid + 1):
        prev = curr

        if i < m and (j >= n or nums1[i] <= nums2[j]):
            curr = nums1[i]
            i += 1
        else:
            curr = nums2[j]
            j += 1

    # If total length is odd -> middle element
    if total % 2 == 1:
        return curr
    # If even -> average of two middle elements
    else:
        return (prev + curr) / 2
```

```python
nums1 = [1, 2]
nums2 = [3, 4]
print(findMedianSortedArrays_simple(nums1, nums2))
```
