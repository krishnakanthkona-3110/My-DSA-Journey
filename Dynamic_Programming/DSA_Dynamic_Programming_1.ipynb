# 1Q). Finding the 6th Fibonacci number:

# 1. Recursive approach

```python
def fibo(n):
  if n == 0:
    return 0

  if n == 1:
    return 1

  return fibo(n-1) + fibo(n-2)
```

#2.  Dynamic Programming approach

Using a list to answer without recursion

Tabulation


```python
def fibonacci(n):
  if n==0:
    return 0
  if n==1:
    return 1

  dp = [0]*(n+1)
  dp[0] = 0
  dp[1] = 1

  for i in range(2,n+1):
    dp[i] = dp[i-1] + dp[i-2]

  return dp[n]
```

```python
fibonacci(60)
```

Type	Complexity

Time	O(n)

Space	O(n)

# 3. Iterative Approach (Best for interviews)

```python
def iter_fibo(n):
  if n == 0:
    return 0
  if n == 1:
    return 1

  prev2 = 0
  prev1 = 1

  for i in range(2, n+1):
    curr = prev1 + prev2
    prev2 = prev1
    prev1 = curr

  return curr
```

```python
iter_fibo(60)
```

Type	Complexity

Time	O(n)

Space	O(1)

You are climbing a staircase and it takes A steps to reach the top.


Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

```python
# Recursive approach

def rec_stair(n):
  if n==1:
    return 1
  if n==2:
    return 2

  return rec_stair(n-1) + rec_stair(n-2)
```

# 3Q) Minimum Number of Squares

Given an integer A. Return minimum count of numbers, sum of whose squares is equal to A.



```python
def recursive_count(n):
  if n==0:
    return 0
  if n==1:
    return 1

  min_f = n
  i =1
  while n- (i*i) >= 0:
    ans = recursive_count(n-(i*i))
    min_f = min(ans+1, min_f)
    i+=1

  return min_f

```

```python
recursive_count(20)
```

Time Complexity:

- In the worst case, for each call with input n, the function iterates over all i such that i² ≤ n, which is approximately √n.

- For each iteration, it makes a recursive call with a smaller argument (n - i²).

- Without memoization, this leads to exponential growth because many subproblems are recomputed multiple times.

- Therefore, the time complexity is roughly O(√n) * T(n - i²), which can be approximated as exponential in n, specifically O(2^{√n}) in the worst case.

Space Complexity: O(n)

```python
# tabular DP

def mincount(n):
  dp = [n]*(n+1)

  dp[0] = 0
  dp[1] = 1

  for i in range(2,n+1):
    min_f = 1000
    j = 1
    while i - (j*j) >= 0:
      ans = dp[i-(j*j)]
      min_f = min(ans+1, min_f)
      j += 1

    dp[i] = min_f
  return dp[n]
```

```python
mincount(5)
```

Time Complexity:

- The outer loop runs from 2 to n, so it executes approximately n-1 times.

- For each iteration i, the inner while loop iterates over all j such that j*j ≤ i. The maximum j for each i is roughly √i.

- Therefore, the total number of inner loop iterations over all i is approximately:

Sum_{i=2}^{n} √i ≈ ∫_{1}^{n} √x dx = (2/3) * n^{3/2}

- Consequently, the overall time complexity is O(n^{3/2}).

```python
def numsquares(n):
  dp = [n]*(n+1)
  dp[0] = 0
  dp[1] = 1

  for target in range(2,n+1):
    for s in range(2,target+1):
      square = s*s

      if target - square < 0:
        break

      dp[target] = min(dp[target], 1+ dp[target - square])
  return dp[n]
```

```python
numsquares(10)
```

Time Complexity:
- The outer loop runs from 2 to n, so it executes approximately n-1 times.

- The inner loop runs from 2 up to target+1, which in the worst case is about n times for each target.

- Inside the inner loop, the operations are constant time, except for the break condition.

- Overall, the nested loops lead to a time complexity roughly O(n^2).

Space Complexity: O(n)


```python
def numSquares(n: int) -> int:

        cache = [n] * (n + 1)
        cache[0] = 0

        for i in range(1, n + 1):
            for s in range(1, i + 1):
                square = s * s
                if i - square < 0:
                    break
                if 1 + cache[i - square] < cache[i]:
                    cache[i] = 1 + cache[i - square]

        return cache[n]
```

```python
numSquares(10)
```
